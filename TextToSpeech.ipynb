```python
from gtts import gTTS

# Create text-to-speech object
text = "Welcome Dr. Hare ram singh."
language = "en"  # English

tts = gTTS(text=text, lang=language, slow=False)

# Save to file
tts.save("output.mp3")

print("Audio saved successfully!")
```

```python
from IPython.display import Audio

# Play audio
Audio("output.mp3")
```

```python
import soundfile as sf
```

```python
import librosa

# Load audio
y, sr = librosa.load("output.mp3")

# Lower pitch (negative values → deeper voice)
y_shifted = librosa.effects.pitch_shift(y, sr=sr, n_steps=-4)

# Save new audio
sf.write("male_output.wav", y_shifted, sr)
```

```python
from IPython.display import Audio

# Play audio
Audio("male_output.wav")
```
